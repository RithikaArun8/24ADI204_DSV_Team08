```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/DataCoSupplyChainDataset.csv', encoding='latin1')

df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.histplot(df['Days for shipping (real)'], bins=7, kde=True)

plt.title('Distribution of Actual Shipping Days')
plt.xlabel('Days for Shipping')
plt.ylabel('Number of Orders')
plt.show()
```

```python
sns.countplot(x='Late_delivery_risk', data=df)

plt.title('Late Delivery Risk Distribution')
plt.xlabel('Late Delivery Risk (0 = No, 1 = Yes)')
plt.ylabel('Number of Orders')
plt.show()
```

```python
sns.countplot(x='Shipping Mode', data=df)

plt.title('Distribution of Shipping Modes')
plt.xlabel('Shipping Mode')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45)
plt.show()
```

```python
sns.countplot(
      x='Category Name',
      data=df,
      order=df['Category Name'].value_counts().index
      )

plt.title('Distribution of Product Categories')
plt.xlabel('Product Category')
plt.ylabel('Number of Orders')
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.barplot(data=df, x='Category Name', y='Sales')
plt.xticks(rotation=45)
plt.title('Sales by Product Category')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(data=df, x='Shipping Mode', y='Sales')
plt.title('Sales by Shipping Mode')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.barplot(data=df, x='Category Name', y='Order Profit Per Order')
plt.xticks(rotation=45)
plt.title('Profit by Product Category')
plt.show()
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(data=df, y='Sales')
plt.title('Outlier Analysis of Sales')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(data=df, y='Order Profit Per Order')
plt.title('Outlier Analysis of Order Profit Per Order')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='Days for shipment (scheduled)', y='Days for shipping (real)')
plt.title('Actual Shipping Days vs Scheduled Shipping Days')
plt.xlabel('Scheduled Shipping Days')
plt.ylabel('Actual Shipping Days')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='Product Price', y='Sales')
plt.title('Sales vs Product Price')
plt.xlabel('Product Price')
plt.ylabel('Sales')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='Order Item Quantity', y='Sales')
plt.title('Sales vs Order Quantity')
plt.xlabel('Order Item Quantity')
plt.ylabel('Sales')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='Late_delivery_risk')
plt.title('Late Delivery Risk Distribution')
plt.xlabel('Late Delivery Risk')
plt.ylabel('Number of Orders')
plt.show()
```
