```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")

# Display plots inside notebook
%matplotlib inline
```

```python
df = pd.read_csv("DataCoSupplyChainDataset.csv",encoding="latin1")

print("Dataset loaded successfully!")
```

```python
print("First 5 rows:")
display(df.head())
```

```python
print("Last 5 rows:")
display(df.tail())
```

```python
print("Shape of dataset:")
print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])
```

```python
print("Column names:")
for i, column in enumerate(df.columns, 1):
    print(i, ".", column)
```

```python
print("Dataset information:")
df.info()
```

```python
print("Data types:")
display(df.dtypes)
```

```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
display(missing_values)


# Only columns having missing values
print("Columns containing missing values:")
display(missing_values[missing_values > 0])
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_table = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": missing_percentage
})

print("Missing value analysis:")
display(missing_table[missing_table["Missing Values"] > 0])
```

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
print("Descriptive statistics for numerical columns:")
display(df.describe())
```

```python
print("Descriptive statistics for all columns:")
display(df.describe(include="all"))
```

```python
unique_values = df.nunique()

print("Number of unique values in each column:")
display(unique_values)
```

```python
for column in df.columns:
    print("\n----------------------------------------")
    print("Column:", column)
    print("Number of unique values:", df[column].nunique())
    print("Sample unique values:")
    print(df[column].unique()[:10])
```

```python
numerical_columns = df.select_dtypes(include=np.number).columns
categorical_columns = df.select_dtypes(include=["object", "category", "bool"]).columns

print("Number of numerical columns:", len(numerical_columns))
print("Numerical columns:")
print(list(numerical_columns))

print("\nNumber of categorical columns:", len(categorical_columns))
print("Categorical columns:")
print(list(categorical_columns))
```

```python
constant_columns = [
    column for column in df.columns
    if df[column].nunique() <= 1
]

print("Constant columns:")
print(constant_columns)
```

```python
print("Columns with very high number of unique values:")

high_cardinality = df.nunique().sort_values(ascending=False)

display(high_cardinality.head(15))
```

```python
print("Distribution of numerical columns:")

for column in numerical_columns:

    plt.figure(figsize=(8, 4))

    sns.histplot(
        data=df,
        x=column,
        kde=True
    )

    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Frequency")

    plt.show()
```

```python
numerical_columns = df.select_dtypes(include=np.number).columns

for column in numerical_columns:
    plt.figure(figsize=(8, 4))
    sns.boxplot(x=df[column].dropna())
    plt.title(f"Boxplot of {column}")
    plt.xlabel(column)
    plt.show()
```

```python
print("Value counts for categorical columns:")

for column in categorical_columns:

    print("\n======================================")
    print(column)
    print("======================================")

    display(df[column].value_counts().head(15))
```

```python
correlation = df[numerical_columns].corr()

plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Matrix")

plt.show()
```

```python
corr_matrix = df[numerical_columns].corr()

corr_pairs = (
    corr_matrix
    .where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
    .stack()
    .sort_values(ascending=False)
)

print("Top positive correlations:")
display(corr_pairs.head(10))

print("Top negative correlations:")
display(corr_pairs.tail(10))
```

```python
outlier_summary = []

for column in numerical_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    outlier_summary.append({
        "Column": column,
        "Outlier Count": len(outliers),
        "Outlier Percentage": round(
            len(outliers) / len(df) * 100, 2
        )
    })

outlier_df = pd.DataFrame(outlier_summary)

print("Outlier analysis:")
display(outlier_df.sort_values(
    "Outlier Count",
    ascending=False))
```

```python
print("Negative values in numerical columns:")

for column in numerical_columns:

    negative_count = (df[column] < 0).sum()

    if negative_count > 0:
        print(column, ":", negative_count)
```

```python
print("Zero values in numerical columns:")

for column in numerical_columns:

    zero_count = (df[column] == 0).sum()

    if zero_count > 0:
        print(column, ":", zero_count)
```

```python
summary = pd.DataFrame({
    "Data Type": df.dtypes,
    "Non-Null Count": df.notnull().sum(),
    "Missing Count": df.isnull().sum(),
    "Missing %": (df.isnull().sum() / len(df) * 100).round(2),
    "Unique Values": df.nunique(),
    "Duplicate Values": df.duplicated().sum()
})

print("Complete column summary:")
display(summary)
```
